```python
import pandas as pd
import matplotlib.pyplot as plt

#Dataset de clasificacion para la deteccion de la diabetes
df_diabetes = pd.read_csv("diabetes_prediction_dataset.csv")

#Dataset de habitos con telefonos  por la noche para regresion o clasificacion
df_sueno = pd.read_csv("bedtime_screentime_sleep_debt.csv")

#Simulación de estabilidad y control de la red eléctrica para regresion o clasificacion
df_redelectrica = pd.read_csv("electrical_grid_stability.csv")

#Aprobacion de creditos bancarios para clasificacion
df_credito = pd.read_csv("loan_approval_dataset.csv")

#Predicion del tiempo para clasificaion
df_tiempo = pd.read_csv("weather_forecast_data.csv")
```

### Valores nulos
Como podemos ver ningunos de los Data frames contienen valores nulos asi que no se necesitan mas analisis en este sentido:

```python
print(f"Diabetes {df_diabetes.isnull().sum().sum()}\n")
print(f"Sueño {df_sueno.isnull().sum().sum()}\n")
print(f"Red electrica {df_redelectrica.isnull().sum().sum()}\n")
print(f"Creditos {df_credito.isnull().sum().sum()}\n")
print(f"Tiempo {df_tiempo.isnull().sum().sum()}\n")
```

### Aqui comprobamos el balanceo de datos de la variable objetivo de cada dataset.
- Lo vemos de diferentes formas primero en numeros brutos

```python
print(df_diabetes["diabetes"].value_counts(), end="\n\n")
print(df_sueno["sleep_debt_category"].value_counts(), end="\n\n")
print(df_redelectrica["stabf"].value_counts(), end="\n\n")
print(df_credito[" loan_status"].value_counts(), end="\n\n")
print(df_tiempo["Rain"].value_counts(), end="\n\n")
```

- Ahora lo vemos en porcentaje

```python
print(df_diabetes["diabetes"].value_counts(normalize=True), end="\n\n")
print(df_sueno["sleep_debt_category"].value_counts(normalize=True), end="\n\n")
print(df_redelectrica["stabf"].value_counts(normalize=True), end="\n\n")
print(df_credito[" loan_status"].value_counts(normalize=True), end="\n\n")
print(df_tiempo["Rain"].value_counts(normalize=True), end="\n\n")
```

- Por ultimo vamos a ver cada data frame en graficos:

```python
df_diabetes["diabetes"].value_counts().plot(kind="bar")
plt.title("Distribución de diabetes")
plt.xlabel("diabetes")
plt.ylabel("pacientes")
plt.show()
```

```python
df_sueno["sleep_debt_category"].value_counts().plot(kind="bar")
plt.title("Calidad del sueño")
plt.xlabel("categorias")
plt.ylabel("pacientes")
plt.show()
```

```python
df_redelectrica["stabf"].value_counts().plot(kind="bar")
plt.title("Red electrica")
plt.xlabel("Estabilidad")
plt.ylabel("simulaciones")
plt.show()
```

```python
df_credito[" loan_status"].value_counts().plot(kind="bar")
plt.title("Concesion de creditos")
plt.xlabel("apto/no atpo")
plt.ylabel("clientes")
plt.show()
```

```python
df_tiempo["Rain"].value_counts().plot(kind="bar")
plt.title("Prevision del lluvia")
plt.xlabel("Tiempo")
plt.ylabel("dias")
plt.show()
```
